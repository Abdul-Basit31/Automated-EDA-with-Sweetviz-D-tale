```python
import pandas as pd
# Load a sample dataset
dataset = pd.read_csv('matches.csv')
```

```python
dataset
```

```python
df=dataset
```

```python
import sweetviz as sv
# Analyze the dataframe
report = sv.analyze(df)
# Display in Colab
report.show_notebook()
# To save as HTML:
report.show_html("sweetviz_report.html")
```

```python
import dtale
import dtale.app as dtale_app

# Set D-Tale to use the Colab notebook settings
dtale_app.USE_COLAB = True

# Start D-Tale
d = dtale.show(df)
d
```

```python
from ydata_profiling import ProfileReport
# Generate report
profile = ProfileReport(df, title="IPL Data", explorative=True)
# Save as an HTML file in the Colab file system
profile.to_file("ydata.html")
```

```python
from autoviz.AutoViz_Class import AutoViz_Class
AV = AutoViz_Class()

# If using a DataFrame (dfte), set filename to an empty string ""
# If using a CSV, set dfte to None
filename = "matches.csv"
dft = AV.AutoViz(
    filename,
    sep=',',
    depVar='survived',
    dfte=df,
    header=0,
    verbose=1,
    lowess=False,
    chart_format='html' # 'html' is often more stable in Colab than 'svg'
)
```
